### implementasi bubble sort

```python
def bubble_sort(arr):
  n = len(arr)
  for i in range(n - 1):
    for j in range(n - i - 1):
      if arr[j] > arr[j + 1]:
        arr[j], arr[j + 1] = arr[j + 1], arr[j]
  return arr

# contoh penggunaan
data = [64, 34, 25, 12, 22, 11, 90]
sorted_data =bubble_sort(data.copy()) # menggunakan copy agar data asli tidak berubah
print("Data sebelum sorting:", data)
print("Data setelah Bubble Sort:", sorted_data)

# OUTPUT
# Data sebelum sorting: [64, 34, 25, 12, 22, 11, 90]
# Data setelah Bubble Sort: [11, 12, 22, 25, 34, 64, 90]
```

### implementasi selection sort

```python
def selection_sort(arr):
  n = len(arr)
  for i in range(n - 1):
    min_index = i
    for j in range(i + 1, n):
      if arr[j] < arr[min_index]:
        min_index = j
      arr[i], arr[min_index] = arr[min_index], arr[i]
  return arr

# contoh penggunaan
data = [64, 25, 12, 22, 11]
sorted_data = selection_sort(data.copy())
print("Data sebelum sorting:", data)
print("Datasetelah Selection Sort:", sorted_data)

# OUTPUT
# Data sebelum sorting: [64, 25, 12, 22, 11]
# Datasetelah Selection Sort: [11, 22, 12, 25, 64]
```

### implementasi insertion sort

```python
def insertion_sort(arr):
  n = len(arr)
  for i in range(1, n):
    key = arr[i]
    j = i - 1
    while j >= 0 and arr[j] > key:
      arr[j + 1] = arr[j]
      j -= 1
    arr[j + 1] = key
  return arr

# contoh penggunaan
data = [64, 34, 25, 12, 22, 11, 90]
sorted_data = insertion_sort(data.copy())
print("Data sebelum sorting:", data)
print("Datasetelah Insertion Sort:", sorted_data)

# OUTPUT
# Data sebelum sorting: [64, 34, 25, 12, 22, 11, 90]
# Datasetelah Insertion Sort: [11, 12, 22, 25, 34, 64, 90]
```

### implementasi merge sort

```python
def merge_sort(arr):
  if len(arr) > 1:
    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]

    merge_sort(left_half)
    merge_sort(right_half)

    i = j = k = 0

    while i < len(left_half) and j < len(right_half):
      if left_half[i] < right_half[j]:
        arr[k] = left_half[i]
        i += 1
      else:
        arr[k] = right_half[j]
        j += 1
      k += 1

    while i < len(left_half):
      arr[k] = left_half[i]
      i += 1
      k += 1

    while j < len(right_half):
      arr[k] = right_half[j]
      j += 1
      k += 1

  return arr

# contoh penggunaan
data = [64, 25, 12, 22, 11]
sorted_data = merge_sort(data.copy())
print("Data sebelum sorting:", data)
print("Data setelah Merge Sort:", sorted_data)

# OUTPUT
# Data sebelum sorting: [64, 25, 12, 22, 11]
# Data setelah Merge Sort: [11, 12, 22, 25, 64]
```

### implementasi quick sort

```python
def quick_sort(arr):
  if len(arr) <= 1:
    return arr
  pivot = arr[len(arr) // 2]
  left = [ x for x in arr if x < pivot]
  middle = [x for x in arr if x == pivot]
  right = [x for x in arr if x > pivot]

  return quick_sort(left) + middle + quick_sort(right)

# contoh penggunaan
data = [64, 25, 12, 22, 11]
sorted_data = quick_sort(data.copy())
print("Data sebelum sorting:", data)
print("Data setelah Quick Sort:", sorted_data)

# OUTPUT
# Data sebelum sorting: [64, 25, 12, 22, 11]
# Data setelah Quick Sort: [11, 12, 22, 25, 64]
```

### LATIHAN TUGAS

```python
def bubble_sort(nama, nilai):
    n = len(nilai)
    for i in range(n - 1):
        for j in range(n - i - 1):
            if nilai[j] < nilai[j + 1]:
                nilai[j], nilai[j + 1] = nilai[j + 1], nilai[j]
                nama[j], nama[j + 1] = nama[j + 1], nama[j]
    return nama, nilai

# Data mahasiswa
nama = ["Messi", "Halland", "Samuel", "Arjuno", "Elino"]
nilai = [70, 96, 83, 65, 92]

print("Data sebelum sorting:", nilai)

sorted_nama, sorted_nilai = bubble_sort(nama.copy(), nilai.copy())

print("Data setelah Bubble Sort:")
for i in range(len(sorted_nama)):
    print("Peringkat", i + 1, ":", sorted_nama[i], "\t=", sorted_nilai[i])

# OUTPUT:
# Data sebelum sorting: [70, 96, 83, 65, 92]
# Data setelah Bubble Sort:
# Peringkat 1 : Halland 	= 96
# Peringkat 2 : Elino 	= 92
# Peringkat 3 : Samuel 	= 83
# Peringkat 4 : Messi 	= 70
# Peringkat 5 : Arjuno 	= 65
```
