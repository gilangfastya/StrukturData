### Queue dengan List

```python
# python program to demonstrate queue implementation using list
queue = []

# append() function to enqueue element in the queue
queue.append('a')
queue.append('i')
queue.append('u')
queue.append('e')
queue.append('o')

print('Initial queue')
print(queue)

# pop() function to dequeue element form queue in FIFO order
print('\nElements dequeued from queue :')
print(queue.pop(0))
print(queue.pop(0))
print(queue.pop(0))

print('\nQueue after removing elements')
print(queue)

# uncomenting print(queue.pop()) will cause an IndexError as the queue is noe empty
```

### Queue dengan Collections.deque

```python
# python program to demonstrate queue implementation using collections.deque
from collections import deque

queue = deque()

# append() function to enqueue element in the queue
queue.append('a')
queue.append('i')
queue.append('u')
queue.append('e')
queue.append('o')

print('Initial queue')
print(queue)

# popLeft() function to dequeue element form stack in LIFO order
print('\nElements dequeued from queue :')
print(queue.popleft())
print(queue.popleft())
print(queue.popleft())

print('\nQueue after removing elements')
print(queue)

# uncomenting print(queue.pop()) will cause an IndexError as the queue is noe empty
```

### Queue dengan queue.LifoQueue

```python
# python program to demonstrate queue implementation using queue module
from queue import Queue

# Initializing a queue
qu = Queue(maxsize=5)

# qsize() show the number of elements in the queue
print(qu.qsize())

# put() function to enqueue element in the queue
qu.put('a')
qu.put('i')
qu.put('u')
qu.put('e')
qu.put('o')

print("Full : ", qu.full())
print("Size : ", qu.qsize())

# get() function to dequeue element form queue
print("\nElements dequeue from the queue")
print(qu.get())
print(qu.get())
print(qu.get())

print("\nEmpty : ", qu.empty())
print("Full : ", qu.full())
```

### algoritma pencarian

Linear Search

```python
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i      # ketemu, kembalikan index
    return -1            # tidak ditemukan

data = [5, 3, 8, 1, 9]
print(linear_search(data, 8))
print(linear_search(data, 7))
```

Binary Search

```python
def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if   arr[mid] == target: return mid
        elif arr[mid] < target:  low  = mid + 1
        else:                    high = mid - 1
    return -1

data = [1, 3, 5, 7, 9, 11]  # harus terurut!
print(binary_search(data, 7))
print(binary_search(data, 6))
```

### algoritma pengurutan

Bubble Sort

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr

print(bubble_sort([64, 34, 25, 12, 22]))
```

Selection Sort

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if arr[j] < arr[min_idx]: min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

print(selection_sort([29, 10, 14, 37, 13]))
```

Insertion Sort

```python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]; j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j+1] = arr[j]; j -= 1
        arr[j+1] = key
    return arr

print(insertion_sort([12, 11, 13, 5, 6]))
```
