```python
queue = []

queue.append('jot')
queue.append('aloy')
queue.append('tepe')

print(queue) # antrian

queue.pop(0) # dequeue

print(queue)
```

```python
from collections import deque

queue = deque()

queue.append('jot')
queue.append('aloy')
queue.append('tepe')

print(queue)

queue.popleft()

print(queue)
```

```python
from queue import Queue

antrian = Queue()

antrian.put('jot')
antrian.put('aloy')
antrian.put('tepe')

print(antrian.get())
print(antrian.get())
print(antrian.get())
```
