# MATERI

### 1. Representasi graf dengan adjacency list

```python
# membuat graf tak berarah menggunakan adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E'],
}

# menanpilkan graf
def print_graph(graf):
  for node in graf:
    print(f"{node} -> {graf[node]}")

print_graph(graph)
```

### 2. Traversal Graf menggunakan BFS

```python
from collections import deque

def bfs(graph, start):
  visited = set()
  queue = deque([start])

  while queue:
    vertex = queue.popleft()
    if vertex not in visited:
      print(vertex, end=" ")
      visited.add(vertex)
      queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

# contoh pemanggilan
bfs(graph, 'A')
```

### 3. Traversal Graf menggunakan DFS

```python
def dfs(graph, start, visited=None):
  if visited is None:
    visited = set()

  if start not in visited:
    print(start, end=" ")
    visited.add(start)
    for neighbor in graph[start]:
      dfs(graph, neighbor, visited)

# contoh pemanggilan
dfs(graph, 'A')
```

### 4. Latihan Personal

```python
def hitung_koneksi(graf):
    hasil = {}
    for simpul in graf:
        hasil[simpul] = len(graf[simpul])
    return hasil

# Data graph
graf_data = {
    'P': ['Q', 'R'],
    'Q': ['P', 'S'],
    'R': ['P', 'T'],
    'S': ['Q'],
    'T': ['R']
}

# Menjalankan fungsi
data_koneksi = hitung_koneksi(graf_data)
print("Jumlah koneksi setiap simpul:")
for simpul, jumlah in data_koneksi.items():
    print(f"Simpul {simpul} mempunyai {jumlah} koneksi")
```

### nomor 1 fungs_graph

```python
graph = {
    'A': ['B', 'C', 'D'],
    'B': ['A', 'C', 'E'],
    'C': ['A', 'B', 'F'],
    'D': ['A', 'E'],
    'E': ['B', 'D', 'F'],
    'F': ['C', 'E'],
}

def print_graph(graph):
    for node in graph:
        print(f"{node} -> {graph[node]}")

print("Struktur Graf")
print_graph(graph)
```

### nomor 2 BFS dan DFS

```python
def bfs(graph, start):
    visited = set()
    queue = deque([start])
    while queue:
        node = queue.popleft()
        if node not in visited:
            print(node, end=" ")
            visited.add(node)
            queue.extend(graph[node])
    print()

def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()
    if start not in visited:
        print(start, end=" ")
        visited.add(start)
        for neighbor in graph[start]:
            dfs(graph, neighbor, visited)

print("\nBFS dari A")
bfs(graph, 'A')

print("\nDFS dari A")
dfs(graph, 'A')
print()
```

### nomor 3

```python
def bfs_list(graph, start):
    visited = set()
    queue = deque([start])
    result = []
    while queue:
        node = queue.popleft()
        if node not in visited:
            result.append(node)
            visited.add(node)
            queue.extend(graph[node])
    return result

print("\nBFS return list")
print(bfs_list(graph, 'A'))
```

### nomor 4

```python
def find_path(graph, start, end, visited=None, path=None):
    if visited is None:
        visited = set()
    if path is None:
        path = []

    visited.add(start)
    path = path + [start]

    if start == end:
        return path

    for neighbor in graph[start]:
        if neighbor not in visited:
            result = find_path(graph, neighbor, end, visited, path)
            if result:
                return result

    return None

print("\nfind_path A -> F")
print(find_path(graph, 'A', 'F'))
```

### nomor 5

```python
def is_connected(graph):
    start = list(graph.keys())[0]
    visited = set()
    stack = [start]
    while stack:
        node = stack.pop()
        if node not in visited:
            visited.add(node)
            stack.extend(graph[node])
    return len(visited) == len(graph)

print("\nis_connected")
print("Graf terhubung?", is_connected(graph))
```
