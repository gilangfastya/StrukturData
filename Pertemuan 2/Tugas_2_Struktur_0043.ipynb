Buatlah double linked list dengan 3 node didalamnya, kemudian lakukan penyisipan tengah dan hapus dengan kata kunci !

```python
class Node:
    def __init__(self, dataval=None):
        self.dataval = dataval
        self.nextval = None
        self.prevval = None


class DoublyLinkedList:
    def __init__(self):
        self.headval = None

    # print the linked list
    def listprint(self):
        printval = self.headval
        while printval is not None:
            print(printval.dataval)
            printval = printval.nextval

    # tambah node di akhir
    def AddInEnd(self, newdata):
        NewNode = Node(newdata)

        if self.headval is None:
            self.headval = NewNode
            return

        last = self.headval
        while last.nextval:
            last = last.nextval

        last.nextval = NewNode
        NewNode.prevval = last

    # penyisipan di tengah
    def InsertMiddle(self, prev_node_data, newdata):
        current = self.headval

        # cari node sebelumnya
        while current is not None:
            if current.dataval == prev_node_data:
                break
            current = current.nextval

        if current is None:
            print("Node tidak ditemukan")
            return

        NewNode = Node(newdata)

        NewNode.nextval = current.nextval
        current.nextval = NewNode
        NewNode.prevval = current

        if NewNode.nextval is not None:
            NewNode.nextval.prevval = NewNode

    # hapus berdasarkan kata kunci
    def RemoveNode(self, key):
        current = self.headval

        while current is not None:
            if current.dataval == key:
                break
            current = current.nextval

        if current is None:
            print("Data tidak ditemukan")
            return

        # node pertama
        if current.prevval is None:
            self.headval = current.nextval
            if self.headval is not None:
                self.headval.prevval = None
        else:
            current.prevval.nextval = current.nextval

        if current.nextval is not None:
            current.nextval.prevval = current.prevval

        current = None

Li = DoublyLinkedList()

# add node
Li.AddInEnd("Januari")
Li.AddInEnd("Februari")
Li.AddInEnd("Maret")

print("Data awal:")
Li.listprint()

# penyisipan tengah setelah Februari
Li.InsertMiddle("Februari", "Februari Tengah")

print("\nSetelah penyisipan tengah:")
Li.listprint()

# hapus berdasarkan kata kunci
Li.RemoveNode("Februari")

print("\nSetelah hapus 'Februari':")
Li.listprint()
```

Buatlah circular linked list dengan 3 node didalamnya, kemudian lakukan penyisipan akhir dan hapus akhir !

```python
class Node:
    def __init__(self, dataval=None):
        self.dataval = dataval
        self.nextval = None


class CircularLinkedList:
    def __init__(self):
        self.headval = None

    # menampilkan isi circular list
    def listprint(self):
        if self.headval is None:
            return

        temp = self.headval
        while True:
            print(temp.dataval)
            temp = temp.nextval
            if temp == self.headval:
                break

    # tambah node di akhir
    def AddInEnd(self, newdata):
        NewNode = Node(newdata)

        # jika kosong
        if self.headval is None:
            self.headval = NewNode
            NewNode.nextval = self.headval
            return

        temp = self.headval
        while temp.nextval != self.headval:
            temp = temp.nextval

        temp.nextval = NewNode
        NewNode.nextval = self.headval

    # hapus node akhir
    def RemoveEnd(self):
        if self.headval is None:
            return

        # 1 node
        if self.headval.nextval == self.headval:
            self.headval = None
            return

        temp = self.headval
        prev = None

        while temp.nextval != self.headval:
            prev = temp
            temp = temp.nextval

        prev.nextval = self.headval
        temp = None

Li = CircularLinkedList()

# add node
Li.AddInEnd("Node 1")
Li.AddInEnd("Node 2")
Li.AddInEnd("Node 3")

print("Data awal:")
Li.listprint()

# penyisipan node akhir
Li.AddInEnd("Node 4")

print("\nSetelah tambah di akhir:")
Li.listprint()

# hapus node akhir
Li.RemoveEnd()

print("\nSetelah hapus akhir:")
Li.listprint()
```
